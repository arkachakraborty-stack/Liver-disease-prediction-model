```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Liver_dataset.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df['Albumin_and_Globulin_Ratio'].fillna(df['Albumin_and_Globulin_Ratio'].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder

encoder=LabelEncoder()

df['Gender']=encoder.fit_transform(df['Gender'])
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
features=['Age','Total_Bilirubin','Direct_Bilirubin','Alkaline_phosphotase','Alamine_ Aminotransferase','Aspartate_Aminotransferase','Total_Proteins','Albumin','Albumin_and_Globulin_Ratio']

scaler=StandardScaler()

df[features]=scaler.fit_transform(df[features])

df.head(3)
```

```python
from imblearn.over_sampling import SMOTE

x=df.drop(['Target','Gender'],axis=1)
y=df['Target']

smote=SMOTE(random_state=42)

X_resampled,y_resampled=smote.fit_resample(x,y)

y_resampled.value_counts()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(X_resampled,y_resampled,test_size=0.2,random_state=42)
```

MODEL SELECTION

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,classification_report

model=LogisticRegression()

model.fit(x_train,y_train)

y_pred=model.predict(x_test)

print("\n\n",classification_report(y_test,y_pred))

print("\n\n",confusion_matrix(y_test,y_pred))
```

RANDOM FOREST

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix,classification_report

model_r=RandomForestClassifier()

model_r.fit(x_train,y_train)

y_pred=model_r.predict(x_test)

print("\n\n",classification_report(y_test,y_pred))
print("\n\n",confusion_matrix(y_test,y_pred))
```

SVC

```python
from sklearn.svm import SVC

model_s = SVC()

model_s.fit(x_train, y_train)

y_pred = model_s.predict(x_test)

print("\n\n",classification_report(y_test,y_pred))
print("\n\n",confusion_matrix(y_test,y_pred))
```
